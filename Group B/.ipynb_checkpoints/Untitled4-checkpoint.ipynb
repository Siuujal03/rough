```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("sales_data_sample.csv", encoding='Latin-1')


df.fillna(df.median(numeric_only=True), inplace=True)

features = ['QUANTITYORDERED', 'PRICEEACH', 'SALES']
X = df[features]

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

df.head()
```

```python
from sklearn.cluster import KMeans

wcss = []
k_range = range(1, 11)
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, marker='o', linestyle='--')
plt.title('Elbow Method')
plt.xlabel('no')
plt.ylabel('wcss')
plt.grid()
plt.xticks(k_range)
plt.show()
```

```python
from sklearn.cluster import KMeans

k_range = range(1, 11)
wcss = []

for k in k_range:
    kmeans = KMeans(n_clusters = k, init='k-means++', n_init=10, random_state = 42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(k_range, wcss, linestyle='--', marker='o')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

```python
kmeans = KMeans(n_clusters = 3, init='k-means++', random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='QUANTITYORDERED', y='SALES', hue='Cluster')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)

cluster_profile = df.groupby('Cluster')[features].mean()

print(cluster_profile)

plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x = 'QUANTITYORDERED', y = 'SALES', hue='Cluster')
plt.show()
```

```python
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

plt.figure(figsize=(30,30))
dendrogram = sch.dendrogram(sch.linkage(X_scaled[:300], method='ward'))
plt.show()

hc = AggolemerativeClustering(n_clusters = 3, metric = 'euclidean', linkage='ward')
df['Hierarchical_Cluster'] = hc.fit_predict(X_scaled)
```
