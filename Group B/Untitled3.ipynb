```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return (x + 3) ** 2
    
def derivative(x):
    return 2 * (x + 3)

x = 2
learning_rate = 0.1
iterations = 1000

for i in range(iterations):
    x = x - learning_rate * derivative(x)


X = np.linspace(-6, 4, 100)
plt.plot(X, f(X), label='y + (x + 3)^2')
plt.scatter(x, f(x), color='red', label='Minimum Point')
plt.legend()
plt.grid()
plt.show()
    
```
