```python
# Group B - Assignment 2
# Problem Statement:
# Classify the email using the binary classification method. Email Spam detection has two states:
# a) Normal State - Not Spam, b) Abnormal State - Spam. 
# Use K-Nearest Neighbors and Support Vector Machine for classification. Analyze their performance.


import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import scale
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn import metrics

print("All required libraries imported successfully.")
```

```python
print("=== Loading Dataset ===")
df = pd.read_csv("emails.csv")

print("\n--- Initial Dataset Head ---")
print(df.head())

print("\n--- Initial Dataset Info ---")
df.info()

print("\n--- Initial Shape ---")
print("Rows:", df.shape[0], "| Columns:", df.shape[1])
```

```python
print("=== Pre-processing Dataset ===")

print("\n--- Missing Values Check ---")
print("Total missing values:", df.isnull().sum().sum())

# Drop missing values if any exist
df.dropna(inplace=True)

# Drop identifier column 'Email No.'
if 'Email No.' in df.columns:
    df.drop(['Email No.'], axis=1, inplace=True)

print("\n--- Shape After Dropping Identifier Column ---")
print("Rows:", df.shape[0], "| Columns:", df.shape[1])

print("\n--- Class Distribution (Target Variable 'Prediction') ---")
print(df['Prediction'].value_counts())
```

```python
print("=== Feature-Target Separation and Scaling ===")

X = df.drop(['Prediction'], axis=1)
y = df['Prediction']

# Feature scaling
X_scaled = scale(X)

print("Features shape (X_scaled):", X_scaled.shape)
print("Target shape (y):         ", y.shape)

# Static Train-Test Split (70:30 ratio)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.30, random_state=42)

print("\nStatic Train-Test Split Complete (70:30 Ratio):")
print("Training set shape (X_train):", X_train.shape)
print("Testing set shape (X_test):  ", X_test.shape)
```

```python
print("=== K-Nearest Neighbors (KNN) Classifier ===")

# Initializing KNN with K=7
knn_model = KNeighborsClassifier(n_neighbors=7)
knn_model.fit(X_train, y_train)

y_pred_knn = knn_model.predict(X_test)

knn_accuracy = metrics.accuracy_score(y_test, y_pred_knn)
knn_cm = metrics.confusion_matrix(y_test, y_pred_knn)

print("\n--- KNN Classification Results ---")
print("KNN Accuracy Score:", knn_accuracy)
print("\nConfusion Matrix:")
print(knn_cm)
print("\nClassification Report:")
print(metrics.classification_report(y_test, y_pred_knn))
```

```python
print("=== Support Vector Machine (SVM) Classifier ===")

# Initializing Linear Support Vector Classifier with C=1.0
svm_model = SVC(C=1.0, kernel='rbf', random_state=42)
svm_model.fit(X_train, y_train)

y_pred_svm = svm_model.predict(X_test)

svm_accuracy = metrics.accuracy_score(y_test, y_pred_svm)
svm_cm = metrics.confusion_matrix(y_test, y_pred_svm)

print("\n--- SVM Classification Results ---")
print("SVM Accuracy Score:", svm_accuracy)
print("\nConfusion Matrix:")
print(svm_cm)
print("\nClassification Report:")
print(metrics.classification_report(y_test, y_pred_svm))
```

```python
print("=== Classifier Performance Comparison ===")
print("--------------------------------------------------")
print(f"{'Classifier Model':<25} | {'Accuracy Score':<15}")
print("--------------------------------------------------")
print(f"{'K-Nearest Neighbors (KNN)':<25} | {knn_accuracy:<15.4f}")
print(f"{'Support Vector Machine (SVM)':<25} | {svm_accuracy:<15.4f}")
print("--------------------------------------------------")
```
