```python
# Group B - Assignment 4
# Problem Statement:
# Implement Gradient Descent Algorithm to find the local minima of a function.
# For example, find the local minima of the function y = (x + 3)^2 starting from the point x = 2.


import numpy as np
import matplotlib.pyplot as plt

print("All required libraries imported successfully.")
```

```python
print("=== Step 1: Define Objective Function and Derivative ===")

# Objective function y = (x + 3)^2
def objective_function(x):
    return (x + 3) ** 2

# First derivative dy/dx = 2 * (x + 3)
def gradient_function(x):
    return 2 * (x + 3)

print("Objective Function: y = (x + 3)^2")
print("Derivative Function: dy/dx = 2*(x + 3)")
```

```python
print("=== Step 2: Implement Gradient Descent Algorithm ===")

# Hyperparameters
cur_x = 2.0           # Starting point x = 2
learning_rate = 0.1   # Step size rate
precision = 0.000001  # Stopping criterion
previous_step_size = 1.0
max_iters = 1000      # Maximum iterations
iters = 0

x_history = [cur_x]
y_history = [objective_function(cur_x)]

print(f"Initial Starting Point x: {cur_x}")
print(f"Learning Rate: {learning_rate}")
print(f"Precision Threshold: {precision}")

# Iterative minimization loop
while previous_step_size > precision and iters < max_iters:
    prev_x = cur_x
    gradient = gradient_function(prev_x)
    
    # Gradient Descent Update Rule
    cur_x = cur_x - learning_rate * gradient
    previous_step_size = abs(cur_x - prev_x)
    
    iters += 1
    x_history.append(cur_x)
    y_history.append(objective_function(cur_x))

print("\n--- Optimization Complete ---")
print(f"Total Iterations Executed: {iters}")
print(f"Calculated Local Minima x: {cur_x:.6f}")
print(f"Minimum Value y:          {objective_function(cur_x):.6f}")
```

```python
print("=== Step 3: Convergence Visualization ===")

x_vals = np.linspace(-6, 3, 100)
y_vals = objective_function(x_vals)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label="y = (x + 3)^2", color='blue')
plt.scatter(x_history, y_history, color='red', s=30, label="Gradient Steps")
plt.plot(x_history, y_history, color='red', linestyle="--", alpha=0.6)

plt.title("Gradient Descent Optimization Trajectory")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```
